#### Przydatne wzory:

Z twierdzenia Bayesa: prawdopodobieństwo zajścia zdarzenia $A$ o ile zajdzie zdarzenie $B$:
$$P(A|B) = \frac{P(B|A)P(A)}{P(B)}$$

Permutacja z powtórzeniami:

$$P_n({k_1, k_2, \ldots, k_m}) = \frac{n!}{k_1!k_2!\ldots k_m!}$$

Permutacja bez powtórzeń:

$$P_n = n!$$

Wariacje z powtórzeniami:

$$W_n^k = n^k$$

Wariacje bez powtórzeń:

$$V_n^k = \frac{n!}{(n-k)!}$$

Kombinacje z powtórzeniami:

$$\overline{C}_{\ n}^{\ k} = {n+k-1 \choose k} = \frac{(k + n - 1)!}{k!(n-1)!}$$

Kombinacje bez powtórzeń:

$$C_n^k = {n \choose k} = \frac{n!}{k!(n-k)!}$$

### Zadanie 1

1) Znajdź minimum funkcji $f(x)$ w przedziale $[-100, 100)$:

$f(x) = \frac{x^{4}}{16} + 5(x + 0.4)^{3} - \frac{x^{2}}{3} - 4 x + 3$

**Podaj wartość $\text{arg min}_xf(x)$ (wartość $x$ dla której funkcja $f(x)$ przyjmuje najmniejszą wartość). oraz wartość _minimum_ ($\text{min}\{f(x)\}$).**

> Uwaga: dopuszczalna tolerancja ± 0.1 dla $x$.

2) Wyświetl wykres funkcji w przedziale $[\text{arg min}_xf(x) - 20, \text{arg min}_xf(x) + 80)$. **Uwaga:** zakres na osi y powinien być tak dopasowany, aby minimalna wartość nie zlewała się z osią X. Punkt odnaleziony punkt w zad. 1 powinien być zaznaczony <font color='red'>czerwoną</font> kropką na wykresie.



````python
import numpy as np
import matplotlib.pyplot as plt

def function(x):
    return (x**4)/16 + 5 * (x + 0.4)**3 - (x**2)/3 - 4*x + 3

X = np.arange(-100, 100, 0.1)
results = []

for x in X:
    results.append(function(x))

# print(function(0))

min_value = min(results)
print(f"Min value: {min_value}")
min_index = results.index(min_value)
x_min = X[min_index]
print(f"X = {x_min}")

print(f"arg min_x f(x): {x_min:.1f}")
print(f"min{{f(x)}}: {min_value:.1f}")

plt.plot(X, results)
plt.ylim([1.1 * min_value, 5000])
plt.xlim((x_min - 20, x_min + 80))
plt.scatter([x_min], function(x_min), color='red')
````

### Zadanie 3

Napisz klasę `MyList`, która:
* reprezentuje listę zawierający maksymalnie 5 elementów,
* elementy można podać jako listę w konstruktorze,
* do listy można dodawać i usuwać elementy,
    * elementy dodawane są na końcu listy,
    * dodanie kolejnego (szóstego) elementu kończy się błędem,
    * elementy można usuwać z $i$-tej pozycji z zachowaniem kolejnosci elementów, które pozostaja na liscie,
* przy usuwaniu wartość usuwanego elementu jest zwracana,
* przy wyświetlaniu obiektu klasy `MyList` wyświetlana jest lista elementów,
* listę można używać jakoś klucz w słowniku,
* dwie listy $a$, $b$, można złączyć za pomocą operatora $+$ pod warunkiem, że łączna liczba elementów obu list nie przekracza limitu pięciu elementów ($|a| + |b| \le 5$).

Przykład:

```python
li_1 = MyList() # utworzenie pustej listy
li_1.append(3)  # dodawanie pojedynczo
li_1.append(7)
li_2 = MyList([2, 5, 8, 1, 0]) # utworzenie poprzez podanie elementow w konstruktorze
#li_2.append(9)  # błąd
li_2.remove(4)  # usuwa piąty element
usuniety = li_2.remove(2)  # usuwa trzeci element
print('wartosc usunietego elementu: ', usuniety)
li_2.remove(2)  # usuwa trzeci element
print('li_1', li_1)
print('li_2', li_2)

d = {
    li_1: 'lista 1',
    li_2: 'lista 2'
}

print(d.keys())

```

Wyjście:
```
wartosc usunietego elementu:  8
li_1 [3, 7]
li_2 [2, 5, 9]
dict_keys([[3, 7], [2, 5, 9], [3, 7, 2, 5, 9]])
```

> Podpowiedź: Przydatne funkcje/metody: `__init__`, `__add__`, `__hash__`, `__str__`, `__repr__`, `hash`, `len`

````python
class MyList:
    def __init__(self, tab=None):
        self.numbers = tab if tab is not None else []
    
    def append(self, num):
        if len(self.numbers) == 5:
            raise ValueError("list is full!")
        self.numbers.append(num)
    
    def remove(self, index):
        if index > len(self.numbers):
            raise ValueError("index out of range!")
        return self.numbers.pop(index)
    
    def __str__(self):
        return f'{self.numbers}'
    
    def __repr__(self):
        return f'{self.numbers}'
    
    def __add__(self, new_list: 'MyList'):
        if len(self.numbers) + len(new_list.numbers) > 5:
            raise ValueError("cannot sum both lists")
        
        return MyList(self.numbers + new_list.numbers)
    

li_1 = MyList() # utworzenie pustej listy
li_1.append(3)  # dodawanie pojedynczo
li_1.append(7)
li_2 = MyList([2, 5, 8, 1, 0]) # utworzenie poprzez podanie elementow w konstruktorze
#li_2.append(9)  # błąd
li_2.remove(4)  # usuwa piąty element
usuniety = li_2.remove(2)  # usuwa trzeci element
print('wartosc usunietego elementu: ', usuniety)
li_2.remove(2)  # usuwa trzeci element
print('li_1', li_1)
print('li_2', li_2)

d = {
    li_1: 'lista 1',
    li_2: 'lista 2'
}

d[li_1 + li_2] = 'lista 2'

print(d.keys())
        
````

Sklep sprzedaje jajka od trzech producentów: A, B, C. Niestety, wystąpiła kurza pandemia, która sprawia, że kury znoszą zakażone jajka. Wiemy, że 10 na 100 jaj jest zakażonych, a producenci dysponują różnymi testami diagnostycznymi do badania jaj. Niżej w tabeli przedstawiono dokładność testów którymi dysponują producenci. Producenci utylizują zakażone jajka, ale chcieliby wiedzieć **jaka jest szansa, że jajo, u którego test wykazał zakażenie, faktycznie jest zakażone?**

Podaj wynik dla każdego producenta i wyświetl w postaci wykresu słupkowego, gdzie na osi X jest nazwa producenta a na osi Y powyższy wynik.

| Producent    |         Dokładność |
| :---         |               ---: |
| Producent A  |              90%   |
| Producent B  |              95%   |
| Producent C  |              99.7% |



````python
import seaborn as sns
import pandas as pd

p_zakazenia = 0.1

dokladnosci = [0.9, 0.95, 0.997]

def faktycznie_zakazone(p_zakazenia, dokladnosc):
    poprawne_zakazenie = p_zakazenia * dokladnosc
    wszystkie_zakazenia = p_zakazenia * dokladnosc + (1 - p_zakazenia) * (1 - dokladnosc)
    return poprawne_zakazenie / wszystkie_zakazenia

producenci = ['A', 'B', 'C']
results = [[producent, faktycznie_zakazone(p_zakazenia, dokladnosc)] for producent, dokladnosc in zip(producenci, dokladnosci)]

results = pd.DataFrame(results, columns=['producent', 'szansa'])
ax = sns.barplot(data=results, y='szansa', x='producent')
ax.bar_label(ax.containers[0], fontsize=10)
````

W wyścigu bierze udział 5 samochodów (autonomicznych). Dwa samochody marki A, dwa samochy marki B i jeden samochód marki C. Zakładając, że każdy model auta ma równe szanse, odpowiedz na pytanie:

Na ile różnych sposobów (z zachowaniem kolejności) może się zakończyć wyścig biorąc pod uwagę tylko marki samochodów (czyli na ile sposobów auta mogą dojechać do mety)? Wypisz wszystkie możliwe zakończenia. Jedno z możliwych zakończeń: ABABC

Zaimplementuj funkcję, która obliczy liczbę możliwych zakończeń wyścigu na podstawie dowolnej liczby aut trzech powyższych producentów (A, B, C). 

Oblicz ile może być zakończeń wyścigu jeżeli weźmie w nim udział 300 samochodów marki A, 60 marki B i 200 marki C (uwaga: będzie to duża liczba, nie wypisuj ani nie tworz listy tych wyników).



````python
from itertools import permutations
from math import factorial

cars = ['A', 'A', 'B', 'B', 'C']
unique_endings = set(permutations(cars))
print(f"Liczba możliwych zakończeń: {len(unique_endings)}")
for ending in sorted(unique_endings):
    print(''.join(ending))

def liczba_zakonczen(nA, nB, nC):
    n = nA + nB + nC
    return factorial(n) // (factorial(nA) * factorial(nB) * factorial(nC))

print("Liczba zakończeń dla 300A, 60B, 200C:", liczba_zakonczen(300, 60, 200))
````
