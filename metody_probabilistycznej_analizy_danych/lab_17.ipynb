### Help Mr. Holmes.
##### Help Mr. Holmes decide what action to take regarding burglary in his house. Take into account at least some of the below information. 
- Assume three possible actions: do nothing, go home, call police. 
- Each action incurs some cost: for “do nothing” it is 0, going home is greater, as it means having to work after hours to make up for lost time. Calling police is still greater, as Mr. Holmes, being at his spare time a PI, does not like to depend on police. 
- If someone (Mr. Holmes or police) appears at the burglary site, he/she may catch a thief. There is much higher chance that the police will catch the thief, as they are in the neighbourhood, while Mr. Holmes must drive form work, which takes significant time. 
- Utility of catching a thief is very, very high (regardless of the fact who catches 
him/her).
